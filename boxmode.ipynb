```python
import numpy as np
from itertools import product

def compute_box_modes(a, b, c, max_index=3, mode_type="all"):
    """
    Compute the resonant frequencies (Hz) of a rectangular cavity of size a x b x c (in meters).
    
    Parameters:
    - a, b, c: dimensions in meters
    - max_index: maximum value of m, n, p to consider
    - mode_type: "all", "TE", or "TM"
    
    Returns:
    - List of tuples: (mode type, m, n, p, frequency in GHz)
    """
    c0 = 299792458  # speed of light in vacuum (m/s)
    result = []

    for m, n, p in product(range(max_index + 1), repeat=3):
        if mode_type == "TE" and m == n == p == 0:
            continue  # no TE000 mode
        if mode_type == "TM" and (m == 0 or n == 0 or p == 0):
            continue  # no TM modes with any 0 index
        if mode_type == "all" and m == n == p == 0:
            continue  # skip 0 mode

        freq = (c0 / 2) * np.sqrt((m / a)**2 + (n / b)**2 + (p / c)**2)
        mode = "TE" if 0 in (m, n, p) else "TM"
        if mode_type == "all" or mode == mode_type:
            result.append((mode, m, n, p, freq / 1e9))  # convert Hz to GHz

    result.sort(key=lambda x: x[-1])  # sort by frequency
    return result

x, y, z = 5e-3, 5e-3, 5e-3  # convert mm to meters
modes = compute_box_modes(x, y, z, max_index=3)

for mode in modes:
    print(f"{mode[0]}{mode[1]}{mode[2]}{mode[3]}: {mode[4]:.3f} GHz")
```
